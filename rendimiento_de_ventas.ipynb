* **Procesos**: Operaciones, cálculos matemáticos y transformaciones lógicas aplicadas sobre las entradas para generar la información requerida (ejemplo: cálculo del promedio y validación de límites).

### 4. Análisis del Problema

El objetivo es determinar si una sucursal cumple con el estándar de ventas de la organización. Para lograrlo, el sistema debe recibir el nombre de la sede y las cifras de tres meses de operación.

La regla de negocio establece que el rendimiento se considera satisfactorio si el promedio trimestral es mayor o igual a **$10,000.00**. 

Para garantizar una estabilidad mínima, se evalúa además que ningún mes individual haya registrado ventas inferiores a **$5,000.00**.

---
### 5. Entradas, Procesos y Salidas

### Entradas

* `nombre_sucursal` (Cadena de texto): Nombre de la sede evaluada.
* `venta_mes1` (Número real): Ventas del primer mes.
* `venta_mes2` (Número real): Ventas del segundo mes.
* `venta_mes3` (Número real): Ventas del tercer mes.

### Procesos

1. Captura de los datos de la sucursal y conversión a formato numérico.
2. Cálculo del promedio de ventas:

$$\text{promedio} = \frac{\text{venta\_mes1} + \text{venta\_mes2} + \text{venta\_mes3}}{3}$$


3. Evaluación de la meta promedio mediante el operador relacional:

$$\text{cumple\_promedio} = \text{promedio} \ge 10000.0$$


4. Verificación de estabilidad mensual mediante operadores relacionales y lógicos:

$$\text{sin\_meses\_criticos} = (\text{venta\_mes1} \ge 5000.0) \text{ and } (\text{venta\_mes2} \ge 5000.0) \text{ and } (\text{venta\_mes3} \ge 5000.0)$$


5. Evaluación global mediante el operador lógico `and`:

$$\text{aprobado} = \text{cumple\_promedio} \text{ and } \text{sin\_meses\_criticos}$$



### Salidas

* Despliegue en pantalla del nombre de la sucursal.
* Promedio mensual obtenido formateado con decimales.
* Estado final del rendimiento (`APROBADO` o `NO APROBADO`).

### 7. Código en Python

````python
"""
Sistema de Evaluación de Rendimiento de Sucursales
Calcula el promedio de ventas trimestral y valida los mínimos esperados.
"""


def evaluar_rendimiento_sucursal():
    # Definición de constantes del negocio
    META_PROMEDIO = 10000.0
    PISO_MINIMO_MENSUAL = 5000.0

    print("=== SISTEMA DE EVALUACIÓN DE SUCURSALES ===")

    # 1. ENTRADAS
    nombre_sucursal = input("Ingrese el nombre de la sucursal: ")
    venta_mes1 = float(input("Ingrese las ventas del mes 1 ($): "))
    venta_mes2 = float(input("Ingrese las ventas del mes 2 ($): "))
    venta_mes3 = float(input("Ingrese las ventas del mes 3 ($): "))

    # 2. PROCESOS
    # Operadores aritméticos: (+) y (/)
    promedio_ventas = (venta_mes1 + venta_mes2 + venta_mes3) / 3.0

    # Operadores relacionales: (>=)
    cumple_promedio = promedio_ventas >= META_PROMEDIO

    # Operadores relacionales y lógicos (and)
    sin_meses_criticos = (
        (venta_mes1 >= PISO_MINIMO_MENSUAL)
        and (venta_mes2 >= PISO_MINIMO_MENSUAL)
        and (venta_mes3 >= PISO_MINIMO_MENSUAL)
    )

    # Evaluación integradora
    resultado_final = cumple_promedio and sin_meses_criticos

    # 3. SALIDAS
    print("\n" + "=" * 50)
    print(f"Sucursal Evaluada : {nombre_sucursal}")
    print(f"Promedio Obtenido : ${promedio_ventas:,.2f}")
    print(f"Meta Promedio     : ${META_PROMEDIO:,.2f}")
    print("-" * 50)

    if resultado_final:
        print("ESTADO: APROBADO")
        print("La sucursal supera la meta promedio y mantiene la estabilidad.")
    else:
        print("ESTADO: NO APROBADO")
        print(
            "La sucursal no alcanza el promedio mínimo o registró meses por debajo de $5,000."
        )


if __name__ == "__main__":
    evaluar_rendimiento_sucursal()
````
